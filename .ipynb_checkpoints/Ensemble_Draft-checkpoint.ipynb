```python
import tensorflow as tf
import os

# --- Updated paths based on your files ---
xception_path = './140K_xception_model.keras'
resnet_path = './140K_resnet50_model.keras'
efficientnet_path = './deepfake (1).h5'
# ------------------------------------------

try:
    xception_model = tf.keras.models.load_model(xception_path)
    resnet_model = tf.keras.models.load_model(resnet_path)
    efficientnet_model = tf.keras.models.load_model(efficientnet_path)

    models = {
        'xception': xception_model,
        'resnet50': resnet_model,
        'efficientnet': efficientnet_model
    }
    print("All models loaded successfully!")

except Exception as e:
    print(f"Error loading models: {e}")
    print("Please ensure the model file paths are correct and the files exist.")
```

```python
def preprocess_image_for_model(img_pil, model_name):
    """Applies the correct preprocessing for a given model."""

    if model_name == 'xception':
        # Target size: (256, 256), Scaling: 1.0/255
        img_resized = img_pil.resize((256, 256))
        img_array = tf.keras.preprocessing.image.img_to_array(img_resized)
        img_array /= 255.0
        return np.expand_dims(img_array, axis=0)

    elif model_name == 'resnet50':
        # Target size: (256, 256), Scaling: 1.0/255
        img_resized = img_pil.resize((256, 256))
        img_array = tf.keras.preprocessing.image.img_to_array(img_resized)
        img_array /= 255.0
        return np.expand_dims(img_array, axis=0)

    elif model_name == 'efficientnet':
        # Target size: (224, 224), Scaling: None (Internal to model)
        img_resized = img_pil.resize((224, 224))
        img_array = tf.keras.preprocessing.image.img_to_array(img_resized)
        # No rescaling needed as it's handled inside the model
        return np.expand_dims(img_array, axis=0)
    else:
        raise ValueError("Unknown model name")
```

```python
import numpy as np
import os
from PIL import Image

# Based on our analysis: 'Fake' is 0, 'Real' is 1.
# A sigmoid output gives the probability of the *positive* class (class 1), which is 'Real'.

def ensemble_predict(img_path):
    """
    Runs prediction on all models and returns the ensemble result.
    Returns the final label, average probability of being 'Real', and individual scores.
    """
    if not os.path.exists(img_path):
        return f"Error: Image path not found at '{img_path}'", None, None

    try:
        img_pil = Image.open(img_path).convert('RGB')
    except Exception as e:
        return f"Error opening image: {e}", None, None

    all_probs = {}

    print("Running predictions...")
    for name, model in models.items():
        # 1. Preprocess the image specifically for this model
        processed_img = preprocess_image_for_model(img_pil, name)

        # 2. Get the prediction
        pred = model.predict(processed_img)[0]

        # 3. Standardize the output probability for the 'Real' class
        # ResNet50 & Xception (binary) output is a single value like [0.98]
        # EfficientNet (categorical) output is like [0.02, 0.98] for (Fake, Real)
        prob_real = pred[1] if len(pred) > 1 else pred[0]

        all_probs[name] = float(prob_real)

    # 4. Average the probabilities (soft voting) for the 'Real' class
    avg_prob_real = np.mean(list(all_probs.values()))

    # 5. Determine the final class based on the average probability
    final_prediction = "Real" if avg_prob_real >= 0.5 else "Fake"

    return final_prediction, avg_prob_real, all_probs
```

```python
from google.colab import files
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

uploaded = files.upload()

# Get the filename of the uploaded file
if len(uploaded.keys()) > 0:
    test_image_path = list(uploaded.keys())[0]
    print(f"\nImage '{test_image_path}' uploaded successfully!")

    # Optional: Display the uploaded image
    img = mpimg.imread(test_image_path)
    plt.imshow(img)
    plt.axis('off')
    plt.show()
else:
    test_image_path = None
    print("No file was uploaded.")
```

```python
if test_image_path:
    final_label, avg_prob, individual_scores = ensemble_predict(test_image_path)

    if final_label and "Error" not in final_label:
        print("\n--- Ensemble Results ---")
        print(f"Final Prediction: {final_label}")
        print(f"Average Confidence (for 'Real'): {avg_prob:.4f}")
        print("\n--- Individual Model Scores (for 'Real') ---")
        for model_name, score in individual_scores.items():
            print(f"{model_name:>15}: {score:.4f}")
    else:
        print(final_label) # This will print an error message if something went wrong.
else:
    print("Please upload an image in the cell above before running the prediction.")
```
